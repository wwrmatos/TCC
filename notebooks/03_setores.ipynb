# Setores censitários e delegacias

Dois plots exploratórios:
1. apenas a malha de setores do `gpkg`;
2. a malha com as delegacias da aba `DPs`.

Os mapas da análise final estão em `02_analise_ssp_dp.ipynb`.

```python
import folium

from tcc.geo import carregar_delegacias, carregar_setores
from tcc.mapas import mapa_base
```

```python
gdf = carregar_setores().reset_index(drop=True)
gdf['id'] = gdf.index.astype(str)
dp = carregar_delegacias()

print('Setores:', len(gdf))
print('Delegacias:', len(dp))
gdf.head()
```

```python
gdf[['CD_SUBDIST', 'NM_SUBDIST']].head(50)
```

## 1. Só os setores

```python
m1 = mapa_base(gdf, zoom_start=9)
folium.GeoJson(
    gdf,
    style_function=lambda x: {
        'fillColor': '#d9d9d9',
        'color': '#333333',
        'weight': 0.3,
        'fillOpacity': 0.6,
    },
    name='Setores',
).add_to(m1)

m1
```

## 2. Setores + delegacias

```python
m2 = mapa_base(gdf, zoom_start=9)
folium.GeoJson(
    gdf,
    style_function=lambda x: {
        'fillColor': '#d9d9d9',
        'color': '#333333',
        'weight': 0.3,
        'fillOpacity': 0.5,
    },
    name='Setores',
).add_to(m2)

for _, row in dp.iterrows():
    folium.CircleMarker(
        location=[row['lat'], row['lon']],
        radius=6,
        color='#c1121f',
        fill=True,
        fill_color='#c1121f',
        fill_opacity=0.95,
        tooltip=row['Delegacia'],
        popup=folium.Popup(
            f"<b>{row['Delegacia']}</b><br>Região: {row['Região Administrativa']}",
            max_width=250,
        ),
    ).add_to(m2)

m2
```
